A small synthetic example of the maintained 2026 modules. Selected Fall 2022 student implementations are in `historical/`.

```python
import numpy as np
from sklearn.datasets import make_circles
from vision import fit_svm,build_model
X,y=make_circles(n_samples=80,factor=.3,noise=.02,random_state=0)
svm=fit_svm(X,y)
print('synthetic training fixture score:',svm.score(X,y))
model=build_model('cnn',3)
print('output shape:',model(np.zeros((1,71,71,3),np.float32)).shape)
```
